### **Aprenderemos a usar las Entradas y Salidas Digitales**

Ya que estas nos permitiran interactuar con el mundo que nos rodea, obteniendo el estado de pulsadores, datos de sensores y en general nos permitiran detectar variaciones en el mundo real para actuar en consecuencia

## ¿Que es una Señal Digital?

A grandes rasgos, ss aquella que solo nos puede propocionar dos estados logicos ALTO y BAJO o en efecto **0** y **1** visto desde el punto de vista digital

Su forma caracteristica es conocida como señal de onda cuadrada



Como dato curioso: Podemos decir que las señales Digitales realmente no existen, ya que esto implica tener un tiempo de subida o de bajada de cero, lo que resulta imposible en un caso real

- Todas las señales fisicas son analogicas, aunque decimos que se comportan como señales digitales, aquellas que se aproximan a dicho comportamiento

En el caso de las entradas o salidas Digitales de nuestra placa ESP32, podemos considerar un estado **Alto cuando el voltaje es de 3,3 voltios** y un estado **Bajo cuando el voltaje es de cero voltios**

### **Entradas Digitales**

Las entradas digitales permiten recibir señales de voltaje, que como vimos en la imagen estaran comprendidas entre los 0 o 3,3 voltios

Las lecturas de estas señales nos sirven para desencadenar reacciones segun la logica que hayamos programado en nuestro circuito o microcontrolador 

**Ej:** La lectura del estado de un boton pulsador, es decir una entrada es capaz de detectar cuando presionamos un pulsador o cuando dejamos de hacerlo



### **Resistencias Pull-Up y Pull-Down**

Sirven para establecer un estado logico en una entrada cuando esta se encuentra en reposo

- **Pull-Up:** Establece un estado High(Alto) en la entrada en estado de reposo

  - Es decir, si usamos una resistencia Pull-Up, cuando no estamos presionando el pulsador, el estado que leera nuestra entrada sera alto y cuando lo presionemos pasara a un estado bajo

- **Pull-Down:** Establece un estado Low(Bajo) en la entrada en estado de reposo

  - Es decir, si usamos una resistencia Pull-Down, cuando no estamos presionando el pulsador, el estado que leera nuestra entrada sera bajo y cuando lo presionemos pasara a un estado alto

**En resumen:** Las resistencias Pull-Up y Pull-Down sirven para evitar falsos estados que se producen por el ruido generado en los circuitos y evitar un estado indeterminado en los pines definidos como entradas

### **Podemos hacer uso de estas resistencias de dos maneras diferentes:**
- **De forma externa:** es decir, conectandolas a la entradas de nuestra placa 
- **De forma Interna:** esta ultima opcion es posible gracias a que el esp32 incorpora en su interior estas resistencias, que podemos activar y usar mediante codigo

Algo que debemos tener en cuenta, es que todos los pines de la placa ESP32 DEV KIT V1 incorporan una resistencia interna, excepto los pines del 34 al 39

- **Cuando usamos resistencias externas:**

```bash
void setup(){
  Serial.begin(9600);
  piMode(pin, INPUT) //definimos el pin en modo de entrada
}
```

- **Cuando usamos las resistencias pulldown internas:** (asi evitamos utilizar componentes adicionales en nuestros circuitos)
(en esta forma, uno de los terminales del pulsador directamente a la entrada y el otro a 3 voltios de la placa)

```bash
void setup(){
  Serial.begin(9600);
  piMode(pin, INPUT_PULLDOWN) //establece el estado de reposo del pulsador en un estado bajo
}
```

Lo habitual es usar resistencias pulldown, aunque existiran casos en los que la logica del circuito requiera de uso de resistencias pullup

Como por ej: Si usamos reles u otros dispositivos que funcionen con una logica de activacion inversa

## **Ahora veremos como usar los Pines como Salidas Digitales**

Las Salidas digitales se podrian decir que son lo contrario a las entradas digitales, pero comparten las mismas caracteristicas en cuanto a estado se refieren

Una salida Digital nos sirve para realizar acciones en el mundo real, como encender un led, activar un rele o hacer sonar un timbre, etc.

Las Salidas Digitales al igual que las Entradas Digitales, comprenden dos estados logicos, alto o bajo, 1 o 0

La placa ESP32 DEV KIT V1 incorpora 20 pines que podemos usar como salidas digitales
