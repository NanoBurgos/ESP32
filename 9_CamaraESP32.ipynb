#### **ESP32-CAM**

Modulo ESP32-CAM
 
Esta basado en el chip ESP32, este modulo incorpora una pequeña camara OV2640 de 2 MP, que podemos reemplazaar por diferentes modelos que se adapten a nuestros proyectos mas exigentes

El modulo cuenta con

- Una ranura para tarjetas microSD
- Un Led Ultra luminoso, que funciona a modo de flash para iluminar cuando se necesite 
- Al estas basado en el chip ESP32, tambien dispone de 10 pines que podran funcionar como entradas o salidas de proposito general
- Un puerto UART para comunicacion serial
- Varias salidas de alimentacion de 3.3 y 5 voltios 

Algo que se hecha en falta en este modulo es un puerto USB para llevar a cabo su programacion, ya que no dispone de uno, necesitamos un conversor TTL USB (este conversor sirve para gestionar la comunicacion USB con nuestras placas a traves del puerto serial)

La camara se puede usar para proyectos que necesiten reconocimiento facial, pero para hacer uso de esta funcionalidad, se debe seleccionar la resolucion en "baja", ya que el procesamiento de esta funcion es alto y con una resolucion mayot, la placa no es capaz de gestionarlo todo, se puede elegir la resolucion 400x296 pixeles 

Una vez cambiada la resolucion, se deben activar las opciones **face detection** y **face recognition**
