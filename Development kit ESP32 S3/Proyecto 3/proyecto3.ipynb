## **Proyecto 3: WS2812 - controlar el LED RGB integrado de la placa**

Para este proyecto es necesario instalar la libreria oficial de Freenove

https://github.com/Freenove/Freenove_Development_Kit_for_ESP32_S3?utm_source=chatgpt.com

y desde el IDE de Arduiono 

```bash
Sketch → Include Library → Add .ZIP Library...
```
ESP32-audioI2S_v2.0.0 es la libreria correcta

Luego, el codigo es: 
```bash
/**********************************************************************
  Filename    : WS2812
  Description : The control board carries ws2812, showing colors such as red, green, blue and white.
  Auther      : www.freenove.com
  Modification: 2023/02/22
**********************************************************************/
#include "Freenove_WS2812_Lib_for_ESP32.h"

#define LEDS_COUNT  1
#define LEDS_PIN    48
#define CHANNEL     0

Freenove_ESP32_WS2812 strip = Freenove_ESP32_WS2812(LEDS_COUNT, LEDS_PIN, CHANNEL, TYPE_GRB);

int m_color[5][3] = { {255, 0, 0}, {0, 255, 0}, {0, 0, 255}, {255, 255, 255}, {0, 0, 0} };
int delayval = 100;

void setup() {
	strip.begin();
	strip.setBrightness(10);	
}
void loop() {
	for (int j = 0; j < 5; j++) {
		for (int i = 0; i < LEDS_COUNT; i++) {
			strip.setLedColorData(i, m_color[j][0], m_color[j][1], m_color[j][2]);
			strip.show();
			delay(delayval);
		}
		delay(500);
	}
}
```

Si verifica sin errores, hacemos upload y vemos si el led de la placa va cambiando de colores

D**ebería ir pasando por:**

rojo → verde → azul → blanco → apagado

No hace falta que veamos el monitor en serie por que este codigo no necesita imprimir nada por serial

**Ahora ya comprobaste otra cosa importante del kit:** el ESP32-S3 puede controlar un LED RGB direccionable usando un único GPIO, en este caso el GPIO 48. Ese tipo de LED no funciona simplemente con HIGH y LOW; recibe datos digitales que indican cuánto rojo, verde y azul debe mostrar.
