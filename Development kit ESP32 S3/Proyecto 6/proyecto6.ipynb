## **Proyecto 6: Crear una interfaz gráfica real: botones, etiquetas, controles**

Ese proyecto ya combina lo que aprendimos en los dos anteriores:

- la pantalla muestra elementos gráficos;
- el touch permite interactuar con ellos.

**Sketch_09_LVGL**

Así que antes de compilar, debemos asegurarnos de usar la versión correcta de LVGL que Freenove incluye. En la carpeta Libraries

```bash
#include "display.h"
#include <lvgl.h>
#include <TFT_eSPI.h>
#include "FT6336U.h"

Display screen;

void setup()
{
    /* prepare for possible serial debug */
    Serial.begin( 115200 );

    /*** Init screen ***/
    screen.init();

    String LVGL_Arduino = "Hello Arduino! ";
    LVGL_Arduino += String('V') + lv_version_major() + "." + lv_version_minor() + "." + lv_version_patch();

    Serial.println( LVGL_Arduino );
    Serial.println( "I am LVGL_Arduino" );

    /* Create simple label */
    lv_obj_t *label = lv_label_create( lv_scr_act() );
    lv_label_set_text( label, LVGL_Arduino.c_str() );
    lv_obj_align( label, LV_ALIGN_CENTER, 0, 0 );

    Serial.println( "Setup done" );
}

void loop()
{
    screen.routine(); /* let the GUI do its work */
    delay( 5 );
}
```
Si compila, puedes hacer Upload directamente. En este proyecto la pantalla debería mostrar una etiqueta centrada parecida a:
```bash
Hello Arduino! V8.3.5
```

Funciona correctamente
