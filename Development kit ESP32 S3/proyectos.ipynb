**Proyectos didacticos con el kit de freenove:**

1. Prueba básica del ESP32-S3 — ya hecha: carga de código + Monitor Serie.
2. Leer el voltaje de alimentación con el ADC — el propio tutorial de Freenove tiene este como primer proyecto.
3. Usar la tarjeta SD — crear archivos, escribir datos, leerlos.
4. Reproducir audio desde la SD usando el módulo amplificador y los dos parlantes. Freenove incluye un proyecto específico para reproducir MP3 desde la SD.
5. Pantalla de 2.8" — mostrar texto, números, imágenes y menús.
6. Pantalla táctil/LVGL — botones, controles y una interfaz gráfica.
7. Visor de imágenes desde la tarjeta SD.
8. Cámara — primero captura, después mostrar la imagen de la cámara directamente en la pantalla.
9. Sensor MAX30102 — leer el sensor y mostrar los valores en Serial y después en pantalla.
10. Reproductor multimedia — pantalla táctil + SD + parlantes. Freenove incluso tiene un ejemplo de reproductor MP3 con LVGL.
11. Wi-Fi — conectarlo a tu red, hacer peticiones HTTP y montar un pequeño servidor web.
12. Proyecto integrador — cámara + pantalla + SD + audio + Wi-Fi en una sola interfaz. Freenove tiene precisamente un ejemplo de “LVGL Multifunctionality” que combina varios de estos elementos.
