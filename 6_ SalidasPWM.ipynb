En la actualidad vivimos rodeados de dispositivos que usan PWM para realizar algunas de sus operaciones, 

Telefonos inteligentes, televisores o la mayoria de electrodomesticos en nuestro hogar hacen uso del PWM

### **¿Que es el PWM?**

**PWM** = **P**ulse **W**idth **M**odulation , traducido al español, Modulacoin por ancho de pulsos

Se trata de una tecnica que se utiliza para transmitir señales basadas en pulsos, estos pulsos nos permiten regular la energia que entregamos a un dispositivo electronico variando rapidamente la energia que este recibe, cambiando entre encendido y apagado

### **Que se puede hacer con el PWM**

Utilizando la tecnica del PWM, podemos realizar multitud de operaciones, como por ej.
- Regular la velocidad de motores
- Modificar la intensidad de lamparas
- Reproducir Sonidos 
- Transmitir informacion a traves de un canal de comunicacion 

### **PWM vs DAC**



El PWM es una tecnica y no debemos confundirlo con las señales analogicas puras(que podiamos generar utilizando el conversor DAC de nuestra placa)

La principal diferencia entre una Salida PWM y una Salida DAC es la resolucion que puede proporcionar cada una de ellas

- El PWM solo nos puede dar dos valores posibles, (0 o 1)

- En cambio el DAC nos puede proporcionar un amplio rango de valores dependiendo de su voltaje y resolucion 

### **Caracteristicas del PWM**

- Es una señal basada en pulsos, y estos pulsos pueden ser altos o bajos 



- El tiempo que la señal permanece en un pulso alto, es conocido como **Duty Cicle**, que traducido al español, significa: ciclo de trabajo



Un ciclo de trabajo o Duty cicle del 0%, dara como resultado la minima luminosidad posible en un foco, es decir, el foco permanecera apagado

En cambio un Duty Cicle de 50% provocara que la intensidad de el foco se incremente hasta la mitad de su luminosidad maxima 
 
Y un Duty Cicle de 100% hara que el foco brille en su totalidad 

De este modo, modificando el ciclo de trabajo, podemos ajustar el nivel de luminosidad de un foco, ajustar la velocidad a la que gira un motor o generar sonidos, entre otras muchas cosas, ya que el ciclo de trabajo determina **la cantidad de energia que entregamos a una carga en un intervalo de tiempo determinado**

### **Como usar el PWM en nuestra placa ESP32**

La placa ESP32  cuenta con 16 canales independientes que pueden configurarse, para generar señales PWM con diferentes propiedades 

Todos los pines que pueden actuar como salidar se pueden usar como pines PWM, excepto los pines del 34 al 39, ya que estos no disponen resistencias Pull-Up o Pull-Down

**Para producir una señal PWM**

Utilizando la placa ESP32, es necesario definir a traves de codigo las siguientes caracteristicas:
- Frecuencia: Cantidad de pulsos por segundo de la señal
- Duty Cicle: Relacion entre el tiempo encendido frente al apagado
- Canal PWM: Uno de los canales disponibles 0 - 15
- GPIO: Uno de los 21 pines disponibles
