## **-----------------------------------------------------------------------------**
## **ESP32 (2016)**

**Resumen**

Fue el primer integrante de la familia y rápidamente reemplazó al ESP8266.

Incluye:

- Wi-Fi
- Bluetooth Classic
- Bluetooth Low Energy (BLE)
- Dos núcleos Xtensa LX6
- Hasta 240 MHz

Es el ESP32 "clásico".

**Está pensado para**
- Domótica
- Automatización
- IoT
- Robots
- Sensores
- Servidores web
- Estaciones meteorológicas
- Proyectos generales

**Ventajas**
- Muy económico
- Muchísima documentación
- Gran comunidad
- Compatible con Arduino IDE, ESP-IDF, PlatformIO

**Desventajas**
- Ya tiene varios años.
- Menor potencia que los modelos modernos.

## **-----------------------------------------------------------------------------**
## **ESP32 S2 (2020)**

**Resumen**

Espressif buscó mejorar la seguridad y agregar USB nativo.

Tiene:

- Un solo núcleo Xtensa LX7
- USB OTG integrado
- Más memoria
- Mejor seguridad

No tiene Bluetooth.

**Está pensado para**
- Dispositivos USB
- Teclados
- Mouse
- Automatización
- Equipos industriales
- Productos comerciales

**Ventajas**
- USB integrado
 -Muy seguro

**Desventajas**
- Sin Bluetooth

## **-----------------------------------------------------------------------------**
## **ESP32 S3 (2021)**

**Resumen**

Actualmente es uno de los modelos más populares.

Mejora mucho el rendimiento respecto al ESP32 original.

**Incluye:**

- Dos núcleos Xtensa LX7
- USB OTG
- Wi-Fi
- Bluetooth LE 5
- Instrucciones vectoriales para IA

**Puede ejecutar algoritmos de:**

- reconocimiento facial
- reconocimiento de voz
- clasificación de imágenes
- Está pensado para
- Visión artificial
- IA en el borde (Edge AI)
- Cámaras
- Interfaces gráficas
- Audio
- Reconocimiento de voz
- Robots

**Ventajas**
- Muy potente
- Mucha memoria
- USB integrado
- Excelente para cámaras

**Desventajas**
- Precio ligeramente mayor

## **-----------------------------------------------------------------------------**
## **ESP32 C3 (2021)**

**Resumen**

Fue el primer ESP32 basado en arquitectura RISC-V.

**Características:**

- Núcleo único RISC-V
- Wi-Fi
- Bluetooth LE 5
- Muy bajo consumo
- Está pensado para
- Sensores IoT
- Dispositivos a batería
- Automatización
- Hogar inteligente
- Wearables

**Ventajas**
- Muy barato
- Consume poca energía
- Excelente para producción masiva

**Desventajas**
- Menor potencia de procesamiento

## **-----------------------------------------------------------------------------**
## **ESP2 H2 (2021)**

**Resumen**

Es un caso especial.

No posee Wi-Fi.

**Tiene solamente:**

- Bluetooth LE
- Zigbee
- Thread

Consume muy poca energía.

**Está pensado para**
- Sensores
- Interruptores inteligentes
- Automatización del hogar
- Redes Mesh

**Ventajas**
- Consumo extremadamente bajo

**Desventajas**
- No sirve si necesitas Wi-Fi.

## **-----------------------------------------------------------------------------**
## **ESP32 C5 (2022)**

**Resumen**

**Es el primer ESP32 con:**

- Wi-Fi 6
- Banda dual

**Puede trabajar en:**

- 2.4 GHz
- 5 GHz

Esto permite reducir interferencias.

**Está pensado para**
- Streaming
- Cámaras IP
- Redes congestionadas
- Automatización industrial

**Ventajas**
- Mayor velocidad
- Menor latencia

**Desventajas**
- Todavía tiene poca disponibilidad en comparación con otros modelos.

## **-----------------------------------------------------------------------------**
## **ESP32 C6 (2023)**

**Resumen**

Es el primer ESP32 con Wi-Fi 6.

**Incluye:**

- RISC-V
- Wi-Fi 6
- Bluetooth LE 5
- IEEE 802.15.4

**Gracias a IEEE 802.15.4 puede utilizar:**

- Thread
- Zigbee
- Matter

**Está pensado para**
- Smart Home
- Automatización industrial
- Domótica moderna
- Redes Mesh
- Matter

**Ventajas**
- Muy preparado para el futuro
- Excelente eficiencia

**Desventajas**
- Aún tiene menor soporte que el ESP32 clásico y el S3.

## **-----------------------------------------------------------------------------**
## **ESP32 P4 (2024)**

**Resumen**

Es el miembro más diferente de la familia.

**No posee:**

- Wi-Fi
- Bluetooth

Está pensado para trabajar junto a otro ESP32 que proporcione la conectividad.

**Posee:**

- Procesador RISC-V de alto rendimiento
- Aceleradores multimedia
- Soporte para pantallas grandes
- Interfaces de alta velocidad

**Está pensado para**
- HMI (Interfaces Hombre-Máquina)
- Pantallas táctiles
- Equipos industriales
- Cámaras
- Multimedia
- Edge Computing
- Robots avanzados

**Ventajas**
- Muchísima potencia
- Ideal para procesamiento local

**Desventajas**
- Necesita otro chip para conectividad inalámbrica.

## **¿Cuál elegir según el proyecto?**

| Tipo de proyecto                                   | Modelo recomendado          | Motivo                                                                                             |
| -------------------------------------------------- | --------------------------- | -------------------------------------------------------------------------------------------------- |
| Aprender ESP32 por primera vez                     | **ESP32 clásico**           | Gran comunidad, abundante documentación y ejemplos.                                                |
| Automatización del hogar (domótica)                | **ESP32-C3** o **ESP32-C6** | Bajo consumo; el C6 añade soporte para Matter, Zigbee y Thread.                                    |
| Robot móvil                                        | **ESP32-S3**                | Mayor potencia de procesamiento y mejores periféricos.                                             |
| Cámara con transmisión de imágenes                 | **ESP32-S3**                | Optimizado para visión artificial y módulos de cámara.                                             |
| Reconocimiento de voz                              | **ESP32-S3**                | Instrucciones vectoriales para IA y procesamiento de audio.                                        |
| Sensores alimentados por batería                   | **ESP32-C3**                | Muy eficiente energéticamente.                                                                     |
| Dispositivos USB (teclados, mouse, HID)            | **ESP32-S2** o **ESP32-S3** | Ambos incluyen USB nativo.                                                                         |
| Redes Zigbee o Thread                              | **ESP32-H2**                | Diseñado específicamente para estos protocolos.                                                    |
| Interfaces gráficas avanzadas y pantallas táctiles | **ESP32-P4**                | Alto rendimiento para multimedia y HMI (requiere un chip adicional para conectividad inalámbrica). |
