### Salidas Analogicas DAC

Este tipo de salidas nos permiten controlar la cantidad de energia que se envia a una carga 

Y podemos utilizarlas por ejemplo:
- Para regular la velocidad de giro de motores
- Regular la intensidad luminosa
- Controlar elementos termoelectricos
- O manejar fuentes conmutadas entre otras muchas cosas


Previamente aprendimos como las Entradas Analogicas eran capaces de interpretar señales conocidas como senoidales o señales analogicas

Ahora, realizaremos lo contrario, en vez de interpretar este tipo de señales, **aprenderemos a generarlas**

### **La placa ESP32 nos permite generar este tipo de señales de maneras diferenntes**, 

- Utilizando su convrrsor DAC (Conversor digital analogico) 

- O haciendo uso de la tecnica PWM (Modulacion de aancho de pulso)

## **¿Que es un conversor digital analogico?**

Se trata de un comversor que permite transformar una señal digital o binaria en una señal de corriente o tension analogica, 

Este tipo de señales nos permiten controlar la cantidad de voltaje que entregamos a una carga, de esta manera podemos regular la intensidad luminosa de lamparas, la velocidad de motores o incluso reproducir sonido entre otras muchas cosas 

La placa ESP32 a diferencia de otros muchos microcontroladores cuenta con un conversor DAC

Este conversor dispone de dos canales independientes y tienen una resolucion de 8 bits 

Estos canales se encuentras en los GPIO´S 25 y 26

Algo que debemos destacar de estas dos salidas analogicas es que son consideradas como **Salidas Analogicas Puras** a diferencia de las salidas PWM que vremos mas adelante


## **----------------------------------------------------------------------------------------------------------------**
El nucleo del ESP32 instalado en el IDE de Arduino nos proporciona una funcion para llevar a cabo el control de estas Salidas Analogicas

Esta Funcion es dacWrite(canal, voltaje)

- **La funcion recibe dos parametros**
  - El primero corresponde con el canal del DAC que queremos controlrar, ya sea el 1 que corresponde al GPIO25 o el 2 que corresponde con el GPIO26
  - El segundo parametro es el nivel de voltaje que queremos entregar por el canal seleccionado y como estos canales disponen de una resolucion de 8 bits, el voltaje sera mapeado de 0 a 255; siendo **0 igual a 0 voltios** y **255 igual a 3.3 voltios**

Es decir, si quisiesemos encender en su totalidad un led, hariamos algo como esto:

```bash
dacWrite(25,255);
```

y si midieramos el voltaje, utilizado un medidor de tension en el pin utilizado, podriamos ver como se aproxima a los 3.3 voltios
