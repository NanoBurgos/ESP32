### Sensores Tactiles del ESP32

La placa ESP32 incorpora 10 sensores tactiles capacitivos

Estos sensores son capaces de detectar variaciones en cualquier cosa que sea conductiva o que tenga una diferencia dielectrica con el aire

Esto se traduce en que cuando acercamos nuestrro dedo a cualquiera de estos sensores, provocamos que una peuqeña corriente fluya a traves de nuestra piel, dando lugar a un cambio en la capacitancia entre el sensor y el entorno

Este cambio es detectado por el ESP32 y de esta manera podemos detectar que el sensor ha sido tocado

#### **Estos 10 sensores tactiles que incorpora el ESP32, podemos encontrarlos en sus siguiente pines** 



Estos son identificados como pines Touch y podemos usarlos utilizando el IDE de Arduino  de una manera muy sencilla 

#### **El nucleo del ESP32 instalado en el entorno de Arduino, nos proporciona una serie de metodos para efectuar la lectura de cualquiera de estos sensores tactiles:**

- **touchRead(Pin):** este metodo recibe como parametro el pin del sensor tactil que queremos monitorizar y nos devuelve un valor numerico relacionado con la capacitancia de dicho sensor

- **touchAttachInterrupt(Pin, Funcion, Umbral)**: utilizando esta funcion podemos adjuntar una interrupcion a un sensor tactil, esta funcion recube tres parametros
  - Pin: corresponde al Pin touch que queremos monitorizar 
  - Funcion: nombre de la funcion que se ejecuitara, cuando la interrupcion tenga lugar 
  - Umbral: umbral que se debe superar para que la interrupcion  se lleve a cabo

- **touchInterruptGetLastStatus(Pin)**: Este devuelve verdadero si el panel tactil ha sido y continua presionado o falso en caso contrario 

- **touchDetachInterrupt(Pin)**: este metodo, recibe como parametro el pin del sensor qeu queremos desvincular 

- **touchInterruptSetThresholdDirection(bool inferior)**: este metodo nos permite configurar si la interrupcion se activara cuando el valor del sensor sea menor o mayor del umbral, de modo a que si queremos un comportamiento diferente al que viene por defecto, debemos llamar a este metodo pasandole como parametro un valor falso

- **touchSleepWakeUpEnable(Pin, Umbral)**: permite despertar del estado de sueño profundo, haciendo uso de sus sensores tactiles

Los sensores tactiles del ESP32 nos permites hacer uso de interrupciones
