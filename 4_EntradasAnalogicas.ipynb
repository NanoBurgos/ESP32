### **Entradas Analogicas**

Este tipo de entradas nos permiten leer el valor de potenciometros, sensores o dispositivos de medicion

## **¿Que es una Señal Analogica?**

Es aquella que puede tomar diferentes valoresd de tension en un perdiodo de tiempo determinado

Su forma caracteristicca es conocida como señal de onda senoidal y su representacion se realiza en una escala de tiempo 



A diferencia de las señales digitales que solo pueden tomar dos valores logicos, las **Señales Digitales** pueden variar en un rango de voltaje determinado

Entonces cuando hablamos de una Entrada Analogica estamos hablando de un pin o terminal que es capaz de interpretar estas señales mediante un microcontrolador

- En el caso de las entradas analogicas de la placa ESP32, el rango de voltaje que podemos leer **oscila entre los 0 y los 3.3 voltios**

La paca de desarrollo ESP32 DEVKIT V1, dispone de 18 pines que podemos usar como entradas analogicas y esto es gracias a que la placa incorpora un conversor Analogico - Digital

- Es decir que estas entradas leeran una señal analogica que despues sera convertida en una señal digital que sera interpretada internamente por el microcontrolador

- Estas entradas tienen una resolucion de 12 bits, esto quiere decir que los valores de 0 a 3.3 voltios **seran convertidos o mapeados internamente a valores comprendidos entre 0 y 4095**, estos seran los valores que usaremos en nuestros programas para establecer logicas en nuestros proyectos

- Algo que debemos tener en cuenta es que estas entradas ADC no tienen un comportamiento lineal, es decir que no podemos diferenciar entre 0 o 0.1 voltios ó entre 3.2 o 3.3 voltios, en otras palabras los pasos entre un voltaje y otro se incrementaran o disminuiran con valores enteros y no decimales

- La resolucion por defecto de este conversor ADC es de 12 bits aunque podremos ajustarla mediante codigo segun nuestras preferencias desde los 9 a los 12 bits

### **Algo importante que debemos saber**

Estos 18 canales que podemos usar como entradas analogicas estas divididos en dos puertos diferentes, el ADC 1 y el ADC 2



Esto es importante ya que el ADC 2 tiene algunas reestricciones que debemos tener en cuenta, y es que resulta que el puerto ADC2 es utilizado por el controlador de wifi de la placa, por lo tanto solo podemos hacer uso del ADC 2 cuando el conrolador de wifi no se ha iniciado
